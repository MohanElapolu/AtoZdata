```python
# Problem: Two Sum
# You are given an integer array nums and an integer target.
# Find the indices of two different elements whose values add up to the target.

# Assumptions:
# Exactly one solution exists, and 
# you may not use the same element twice.
# Return the two indices in any order.
```

```python
# Why does Two Sum ask us to return the indices rather than the actual values?

# Answer:
# The problem asks for indices because 
# the same value can appear multiple times in the array.
# Returning the indices allows us to uniquely identify the elements, 
# even if their values are the same.
```

```python
# Brute Force Approach:
# How would you systematically check 
# the array to find a valid pair without missing any possibilities?

# The brute force approach involves 
# checking all possible pairs of elements in the array to see if they 
# add up to the target. This can be done using two nested loops,
# where the outer loop picks one element and the inner loop checks 
# the remaining elements for a match.

# The time complexity of this approach is O(n^2),
# where n is the number of elements in the array.

# The space complexity is O(1) 
# since we are not using any additional data structures 
# that grow with input size.
```

```python
# Even broadcasting will not necessarily help in this case,
# because we are looking for pairs of indices,
# and broadcasting is more suited for element-wise operations on arrays.

# The time complexity for this also remains O(n^2) 
# and space complexity is O(1).
```

```python
# In solving this we use a hash map 
# to store the elements we have seen so far and their indices.
# In python, we can use a dictionary for this purpose.

# Algorithm:
# 1. Initialize an empty dictionary to store elements and their indices.
# 2. Iterate through the array with index and value.
# 3. For each element, calculate the complement 
#    by subtracting the current element from the target.
# 4. Check if the complement exists in the dictionary.
# 5. If it exists, return the index of the complement and the current index.
# 6. If it does not exist, add the current element 
#    and its index to the dictionary.
```

```python
# Why Why is its average time complexity O(n), 
# even though we perform both a dictionary lookup and 
# an insertion during each iteration?

# Answer:
# The average time complexity is O(n) because
# both dictionary lookups and insertions are O(1) on average.
```

```python
def twoSum(nums, target):
    seen = {}

    for i, num in enumerate(nums):
        complement = target - num

        if complement in seen:
            return [seen[complement], i]

        seen[num] = i

    return []
```

```python
print(twoSum([2, 7, 11, 15], 9))  # [0, 1]
print(twoSum([3, 2, 4], 6))       # [1, 2]
print(twoSum([3, 3], 6))          # [0, 1]
```

```python
# Interview question
# Why is if complement in seen more efficient 
# than if complement in list(seen.keys())?

# Yes when we say, 
# 2 in seen
# it checks the keys
# 2 in seen keys, average complexity O(1)
# 2 in seen.keys(), average complexity O(1)
# 2 in seen.values(), average complexity O(n)
# 2 in list(seen.keys()), average complexity O(n)
```
