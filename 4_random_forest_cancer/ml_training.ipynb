### Training dataset

```python
X = df.drop(columns = ["target"])
y = df["target"]

print("The shape of feature matrix X: {}".format(X.shape))
print("The shape of the target vectory y: {}".format(y.shape))
```

```python
from sklearn.model_selection import train_test_split
```

```python
# train test split at 80-20 ratio
X_train, X_test, y_train, y_test = train_test_split(X,
                                                    y,
                                                    test_size=0.2,
                                                    stratify=y,
                                                    random_state=42,
                                                    )

print("The shape of X train: {}".format(X_train.shape))
print("The shape of y train: {}".format(y_train.shape))
print("The shape of X test: {}".format(X_test.shape))
print("The shape of y test: {}".format(y_test.shape))
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
# Random Forest Classifier training
model = RandomForestClassifier()

model.fit(X_train, y_train)
```

```python
# Model predictions

y_predict = model.predict(X_test)
```

```python
from sklearn.metrics import (recall_score, 
                             precision_score, 
                             f1_score, 
                             confusion_matrix,
                             accuracy_score,
                             classification_report,
                            )
```

```python
a_score = accuracy_score(y_test, y_predict)

print("The accuracy score of the model: {}".format(a_score))
```

```python
r_score = recall_score(y_test, y_predict)
p_score = precision_score(y_test, y_predict)
f_score = f1_score(y_test, y_predict)

print("The recall score of the model: {}".format(r_score))
print("The precision score of the model: {}".format(p_score))
print("The f1 score of the model: {}".format(f_score))
```

```python
print("Confusion Matrix\n")
print("----------------------")
print(confusion_matrix(y_test, y_predict))
print("------------------------------------------")
print("Classification Report\n")
print("------------------------")
print(classification_report(y_test, 
                            y_predict, 
                            target_names=["Malignant", 
                                          "Benign"]
                            ))
```

```python

print("Training accuracy:", model.score(X_train, y_train))
print("Testing accuracy:", model.score(X_test, y_test))
```

```python
# Interpretation:
# - 100% training accuracy: The Random Forest perfectly classified the training samples.
# - 95.61% test accuracy: The model performs well on unseen samples.
# - 4.39% gap: Suggests some overfitting, but not necessarily severe overfitting.
```

```python
# Potential Improvements:
# (Fine Tune below parameters)
# - max_depth, 
# - min_samples,
# - max_features
```

```python
# Lets do Hyper Parameter Tuning
param_grid = {
    'max_depth':[10, 20, 30, None],
    'min_samples_split':[2, 5, 10],
    'min_samples_leaf':[1, 2, 4],
    'max_features':['sqrt', 'log2', None],
    }
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
random_search = RandomizedSearchCV(
    estimator = model,
    param_distributions=param_grid,
    n_iter=300,
    cv=2,
    scoring="recall_weighted",
    n_jobs=-1,
    random_state=42,
    verbose=1,
)
```

```python
random_search.fit(X, y)
```

```python
# 5. Inspect the best parameters and extract the tuned model
print("Best Parameters:", random_search.best_params_)
print("Best CV Score:", random_search.best_score_)

best_rf = random_search.best_estimator_
```

```python
model_sel = RandomForestClassifier(min_samples_split=2, 
                                   min_samples_leaf=1, 
                                   max_features='sqrt', 
                                   max_depth=10,
                                   )
```

```python
model_sel.fit(X_train, y_train)
```

```python
y_predict_new = model_sel.predict(X_test)
```

```python
print("Training accuracy:", model_sel.score(X_train, y_train))
print("Testing accuracy:", model_sel.score(X_test, y_test))
```

```python
print("Confusion Matrix\n")
print("----------------------")
print(confusion_matrix(y_test, y_predict_new))
print("------------------------------------------")
print("Classification Report\n")
print("------------------------")
print(classification_report(y_test, 
                            y_predict_new, 
                            target_names=["Malignant", 
                                          "Benign"]
                            ))
```

#### Recall Macro
When you have a multiclass problem having fairly equal importance for both classes. Use *recall_macro* in *scikit-learn* environments. Computes recall for each class independently and averages them equally. 

#### Recall Weighted
When you have multiclass problem with class imbalance. Use *recall_weighted*, it computes recall for each class and weights the average by class support.
