```python
# Interview Relevance
# Duplicate detection, data validation, and efficient lookups

# Problem
# Input: an integer array nums
# Output: True (if any value appears atleast twice)
# Output: False (if every element is unique)
```

```python
# Brute Force implementation
# - Outer loop: Select an element.
# - Inner loop: Compare it against other elements.
# - Time complexity: O(n^2).
# - Space complexity: O(1)
# we only use loop variables and don't create additional data structures.
```

```python
# Optimized implementation
def duplicate_find(nums):
    if nums:
        seen = set()
        len_num = len(nums)
        for i in range(len_num):
            if nums[i] in seen:
                return True
            else:
                seen.add(nums[i])

    return False
```

```python
# Given nums
duplicate_find([2, 4, 5, 2])
```
