```python
import sqlite3
import pandas as pd
from sklearn.datasets import load_breast_cancer
```

```python
# Load the breast cancer dataset
data = load_breast_cancer(as_frame=True)
df = data.frame.copy()
```

```python
df.info()
```

```python
# Create a SQLite database and connect to it
conn = sqlite3.connect("breast_cancer.db")
```

```python
# Store the dataset
df.to_sql(
    "breast_cancer",
    conn,
    if_exists = "replace",
    index = False,
    )
```

```python
# Verify SQL records
# Write an SQL query to find... 
# Number of malignant samples
# Number of benign samples
print(pd.read_sql_query("SELECT target, COUNT(*) AS total_samples from breast_cancer GROUP BY target", conn))

# Here GROUP BY creates group based on unique target values in a column
# COUNT(*) counts the rows within each group
```

```python
# Write an SQL query to find...
# and display, if the categories are greater than 250
# Here we should use GROUP BY on column: target
print(pd.read_sql_query("SELECT target, count(*) AS total_samples FROM breast_cancer GROUP BY target HAVING total_samples > 250", conn))
```
