```python
import sqlite3

from sklearn.datasets import fetch_california_housing
import pandas as pd
```

```python
# Load California Housing dataset
housing = fetch_california_housing(as_frame=True)
df = housing.frame
```

```python
# Connect to SQLite database
conn = sqlite3.connect("california_housing.db")

# Store dataset as a SQL table
df.to_sql(
    "housing",
    conn,
    if_exists="replace",
    index=False
)
```

```python
# Verify number of records
cursor = conn.execute("SELECT COUNT(*) FROM housing")
print("Total records:", cursor.fetchone()[0])
print("Successfully completed data igestion and storage in SQLite database.")
```

```python
conn.close()
```

```python
# Interview quesiton
# Using the California Housing dataset,
# Write a SQL query
# To idetify the top 5 census blocks groups
# With the highest median house value
"""
SELECT MedInc, HouseAge, Population, MedHouseval
FROM housing
ORDER BY MedHouseVal DESC
LIMIT 5;
"""

# ORDER BY defaults to ascending order.
# To sort in descending order, use the DESC keyworrd.
```

```python
# Interview question
# Using the California Housing dataset,
# Write a SQL query
# To calculate the average median house value
# For each house-age group
# Consider only groups with average median house value greater than 2
"""
SELECT HouseAge, AVG(MedHouseVal) AS AvgMedHouseVal
FROM housing
GROUP BY HouseAge
HAVING AvgMedHouseVal > 2
ORDER BY AvgMedHouseVal DESC;
"""

# WHERE filters individual observations before grouping
# While HAVING filters groups after aggregation
```

```python
# Interview question
# Using the California Housing dataset,
# Write a SQL query
# Calculate the average median house value
# For census block groups with a population greater than 1000

"""
SELECT AVG(MedHouseVal) AS AvgMedHouseVal
FROM housing
WHERE Population > 1000;
"""
```
