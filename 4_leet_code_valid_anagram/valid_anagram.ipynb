```python
# Problem
# Input: Given two strings s and t
# Output: Return true if t is an anagram of s, and 
# Output: False if t is not an anagram of 

# Note: An anagram is a word or phrase 
# contains exactly the same characters of the original string,
# conatins each character with the same frequencies of the original string,
# Anagram, typically uses all the original letters exactly once.

# Example 1:
# Input: s = "anagram", t = "nagaram"
# Output: true

# Example 2:
# Input: s = "rat", t = "car"
# Output: false
```

```python
# Brute Force Approach:
# Create an empty dictionary
# We loop the first string
# Create a key for each string
# Its value is the frequency 
# obtained by incrementing everytime we encounter the word
# We can do the same thing for the second string
# compare the two dictionaries 
# if they are same...we return True
# otherwise...False

# Time complexity O(n+m+k) ~ O(n+m) ~ O(n) [s loop, t loop, comparison loop]
# Space complexity O(26+26) ~ O(1) [s dict can have max 26, t dict can have max 26]
```

```python
def isAnagrams(s, t):
    if len(s) != len(t):
        return False
    dict_s = {}
    dict_t = {}
    for i in range(len(s)):
        if s[i] not in dict_s:
            dict_s[s[i]] = 1
        else:
            dict_s[s[i]] = dict_s[s[i]] +1

        if t[i] not in dict_t:
            dict_t[t[i]] = 1
        else:
            dict_t[t[i]] = dict_t[t[i]] +1
    print(dict_s)
    print(dict_t)
    return dict_s == dict_t
```

```python
s = "samanvi"
t = "saanavi"
isAnagrams(s,t)
```

```python
# For this implementation We achieved
# Length Comparison: O(1)
# Frequency counting loop: O(n)
# Dictionary comparison: O(k)

# Finaly time complexity: O(n+k+1) ~ O(n)

# Space complexity: O(26) + O(26) ~ O(1)
```

```python
# Optional solution increment decrement the dictionary
# Here we use the same dictionary

def isAnagram(s, t):
    if len(s) != len(t):
        return False

    counts = {}

    for char in s:
        counts[char] = counts.get(char, 0) + 1

    for char in t:
        counts[char] = counts.get(char, 0) - 1

    return all(count == 0 for count in counts.values())
```
