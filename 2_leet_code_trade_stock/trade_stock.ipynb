```python
# Trade Stock

# The problem is to find the best time to trade stock.
# Here, trade means buy and sell stock.

# Input
# It is an array of consecutive day stock prices.

# Output
# Maximize profit by finding the best day to buy and sell stock.

# Assumptions
# 1. You must buy before you sell.
# 2. If no profit is possible, return 0.
```

```python
# Brute Force Approach
# We can use nested loops 
# Outer loop [i] to select the day to buy stock
# Inner loop [j] to select the day to sell stock
# Calculate profit = prices[j] - prices[i]
# Track the maximum profit from zero and return it.

# Time Complexity: O(n^2)
# Space Complexity: O(1)
```

```python
# Interview question
# Instead of comparing every pair, suppose we traverse the array only once.
# What two values do we need to track to find the maximum profit?

# 1. The minimum price seen so far (best buying price)
# 2. The maximum profit seen so far (best selling price)
```

```python
# Write the function now

def maxProfit(prices):
    min_price = prices[0]
    profit  = 0

    len_prices = len(prices)
    for i in range(1, len_prices):
        if min_price > prices[i]:
            min_price = prices[i]

        temp_profit = prices[i]-min_price
        if temp_profit > profit:
            profit = temp_profit

    return profit
```

```python
# prices
print(maxProfit([7, 1, 5, 3, 6, 4]))  # 5
print(maxProfit([7, 6, 4, 3, 1]))     # 0
```

```python
# We demonstrated the below things with the algorithm
# 1. Brute force → Optimization: Reduced O(n2) to O(n).
# 2. State tracking: Maintained the minimum historical price and maximum profit.
# 3. Temporal constraints: Ensured the buying price occurred before the selling price.
```
